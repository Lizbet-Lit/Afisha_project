```python
    query = '''
   SELECT 
    p.user_id, 
    p.device_type_canonical,
    p.order_id,
    p.created_dt_msk AS order_dt,  
    p.created_ts_msk AS order_ts,      
    p.currency_code,
    p.revenue,
    p.tickets_count,
    e.event_id,
    p.service_name,
    e.event_type_main,
    c.city_name, 
    r.region_name
FROM afisha.purchases AS p 
LEFT JOIN afisha.events AS e ON p.event_id = e.event_id  
LEFT JOIN afisha.city AS c ON e.city_id = c.city_id
LEFT JOIN afisha.regions AS r ON c.region_id = r.region_id
WHERE    
    p.device_type_canonical IN ('mobile', 'desktop')  
    AND e.event_type_main != 'movie';           
'''
```

```python
df = pd.read_sql_query(query, con=engine)
```

```python
# 1. Сортируем данные по пользователю и времени.
df = df.sort_values(by=['user_id', 'order_ts'])

# 2. Считаем разницу во времени между текущей строкой и предыдущей строкой ВНУТРИ группы пользователя.
time_diff = df.groupby('user_id')['order_ts'].diff()

# 3. Переводим Timedelta в количество дней (float).
df['days_since_prev'] = time_diff.dt.days
```

**Задача 2.1:** Данные о выручке сервиса представлены в российских рублях и казахстанских тенге. Приведите выручку к единой валюте — российскому рублю.

Для этого используйте датасет с информацией о курсе казахстанского тенге по отношению к российскому рублю за 2024 год — `final_tickets_tenge_df.csv`. Его можно загрузить по пути `https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv')`

Значения в рублях представлено для 100 тенге.

Результаты преобразования сохраните в новый столбец `revenue_rub`.

---


```python
# Загрузка информацией о курсе казахстанского тенге по отношению к российскому рублю за 2024 год
url = 'https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv'
rates = pd.read_csv(url)
```

```python
rates.head()
```

```python
rates = rates.rename(columns={'data': 'date'})
rates['date'] = pd.to_datetime(rates['date'])

# В основном датасете берём колонку created_dt_msk и тоже делаем её датой
df['date'] = pd.to_datetime(df['order_dt'])

### Шаг 2: Объединяем таблицы по дате

# left join: берём все строки из df (продаж) и ищем к ним курс по дате
merged = pd.merge(df, rates, on='date', how='left')

### Шаг 3: Считаем выручку в рублях

# Создаём столбец и сразу заполняем его нулями или копируем revenue (как базу)
merged['revenue_rub'] = merged['revenue']

# Теперь перезаписываем только те строки, где валюта KZT
# Формула: выручка * (курс / 100)
mask_kzt = merged['currency_code'] == 'KZT'
merged.loc[mask_kzt, 'revenue_rub'] = merged.loc[mask_kzt, 'revenue'] * (merged.loc[mask_kzt, 'curs'] / 100)
```

```python
# Посмотрим на статистику нового столбца
print(merged['revenue_rub'].describe())

# Проверим, не осталось ли NaN там, где не должно быть
# Например, если у нас был KZT, но не было курса (curs = NaN), то revenue_rub тоже станет NaN
print(merged[merged['currency_code'] == 'KZT']['revenue_rub'].isna().sum())
```

```python
# Находим аномальные строки
anomalies = merged[merged['revenue_rub'] < 0]

# Смотрим, сколько их всего
print(f"Количество транзакций с отрицательной выручкой: {len(anomalies)}")

# Выводим первые 5 строк, чтобы глазами глянуть на детали
print(anomalies.head())
```

```python
# Ищем строки, где и выручка < 0, И количество билетов < 0
refunds_check = merged[(merged['revenue_rub'] < 0) & (merged['tickets_count'] < 0)]
print(f"Строк с отрицательной выручкой И отрицательным количеством билетов: {len(refunds_check)}")
print(refunds_check[['revenue_rub', 'tickets_count', 'currency_code']].head())
```

```python
anomalies = merged[merged['revenue_rub'] < 0]
print(anomalies['revenue_rub'].describe())
print("\nСамые частые значения отрицательной выручки:")
print(anomalies['revenue_rub'].value_counts().head(10))
```

```python
# Создаём признак: 'sale' (продажа), 'adjustment' (корректировка/возврат) или 'other'
def classify_transaction(row):
    if row['revenue_rub'] > 0:
        return 'sale'
    elif row['revenue_rub'] <= 0:
        # Учитывая, что у нас нет отрицательных билетов, считаем всё с минусом корректировками
        return 'adjustment'
    else:
        return 'other'

merged['transaction_type'] = merged.apply(classify_transaction, axis=1)

# Проверяем, сколько чего получилось
print(merged['transaction_type'].value_counts())
```

```python
# 1. Создаем копию исходного датасета
clean_df = df.copy()

# 2. Добавляем колонку
clean_df['revenue_rub'] = merged['revenue_rub']

# Удаляем колонку 'date', она больше не нужна
clean_df = clean_df.drop(columns=['date'])

# Проверяем, что получилось
print(clean_df.info())
```

##  Создание профиля пользователя

В будущем отдел маркетинга планирует создать модель для прогнозирования возврата пользователей. Поэтому сейчас они просят вас построить агрегированные признаки, описывающие поведение и профиль каждого пользователя.

---

**Задача 3.1.** Постройте профиль пользователя — для каждого пользователя найдите:

- дату первого и последнего заказа;
- устройство, с которого был сделан первый заказ;
- регион, в котором был сделан первый заказ;
- билетного партнёра, к которому обращались при первом заказе;
- жанр первого посещённого мероприятия (используйте поле `event_type_main`);
- общее количество заказов;
- средняя выручка с одного заказа в рублях;
- среднее количество билетов в заказе;
- среднее время между заказами.

После этого добавьте два бинарных признака:

- `is_two` — совершил ли пользователь 2 и более заказа;
- `is_five` — совершил ли пользователь 5 и более заказов.

**Рекомендация:** перед тем как строить профиль, отсортируйте данные по времени совершения заказа.

---


```python
# Сортировка 
# Сортируем по пользователю и по времени заказа
clean_df_sorted = clean_df.sort_values(by=['user_id', 'order_dt'])

# Агрегация количественных метрик (сумма, среднее, количество)
numeric_agg = clean_df_sorted.groupby('user_id').agg(
    total_orders=('order_id', 'count'),        
    avg_revenue=('revenue_rub', 'mean'),       
    avg_tickets=('tickets_count', 'mean')        
).reset_index()

# Агрегация качественных метрик
first_order_features = clean_df_sorted.groupby('user_id').agg(
    first_order_date=('order_dt', 'first'),
    first_device=('device_type_canonical', 'first'),
    first_region=('region_name', 'first'),
    first_partner=('service_name', 'first'),
    first_genre=('event_type_main', 'first')
).reset_index()

# Расчёт среднего времени между заказами
# Берем среднее по колонке days_since_prev. 
# Если у пользователя 1 заказ, там будет NaN
avg_gap = clean_df_sorted.groupby('user_id')['days_since_prev'].mean().reset_index()
avg_gap.rename(columns={'days_since_prev': 'avg_days_between_orders'}, inplace=True)

# Сборка профиля (Merge всех частей)
# Начинаем с числовых данных и присоединяем остальные по user_id
user_profile = numeric_agg.merge(first_order_features, on='user_id')
user_profile = user_profile.merge(avg_gap, on='user_id', how='left')

# Добавление бинарных признаков
# is_two: 1, если заказов >= 2, иначе 0
user_profile['is_two'] = (user_profile['total_orders'] >= 2).astype(int)

# is_five: 1, если заказов >= 5, иначе 0
user_profile['is_five'] = (user_profile['total_orders'] >= 5).astype(int)

# Проверка результата
print(f"Итоговый профиль пользователя содержит {user_profile.shape[0]} строк (пользователей)")
print(user_profile.head())
print("\nСтатистика по новым бинарным признакам:")
print(user_profile[['is_two', 'is_five']].sum())
```

```python
# 1. Находим дату первого заказа для каждого пользователя
# Нам нужен минимальный order_ts для каждого user_id
first_orders = df.groupby('user_id')['order_ts'].min().reset_index()
first_orders.columns = ['user_id', 'first_order_ts']

# 2. Присоединяем эту дату обратно к исходному датасету, чтобы найти тип первого мероприятия
df_with_first = pd.merge(df, first_orders, on='user_id', how='left')

# 3. Фильтруем: оставляем только те строки, где дата заказа равна дате первого заказа
# Так мы получим ровно одну строку на пользователя - его самое первое действие
users_first_event = df_with_first[df_with_first['order_ts'] == df_with_first['first_order_ts']].copy()

# 4. Оставляем только нужные колонки: user_id, тип первого мероприятия
# И сразу переименуем колонку типа мероприятия в first_event_type, чтобы не путаться
users_first_event = users_first_event[['user_id', 'event_type_main']].rename(columns={'event_type_main': 'first_event_type'})

# 5. Добавляем признак возврата (returned)
# Считаем общее количество заказов на пользователя
orders_count = df.groupby('user_id')['order_id'].nunique().reset_index()
orders_count.columns = ['user_id', 'total_orders']
orders_count['returned'] = (orders_count['total_orders'] >= 2).astype(int)

# 6. Склеиваем: теперь у каждого пользователя есть "первый тип" и "вернулся ли он"
users_analysis = pd.merge(users_first_event, orders_count[['user_id', 'returned']], on='user_id', how='left')

# Проверка: нет ли пропусков
print("Пропуски в returned:", users_analysis['returned'].isnull().sum())
print("Первые 5 строк таблицы для анализа:")
print(users_analysis.head())
```

```python
# 1. Общее число уникальных пользователей по типу ПЕРВОГО мероприятия
# (В users_analysis каждый пользователь уже встречается 1 раз, поэтому count() достаточно)
users_by_first_event = users_analysis.groupby('first_event_type')['user_id'].count()

# 2. Число вернувшихся пользователей (где returned == 1) по типу ПЕРВОГО мероприятия
returned_users_by_first_event = users_analysis[users_analysis['returned'] == 1].groupby('first_event_type')['user_id'].count()

# 3. Считаем долю
share_returned_first_event = returned_users_by_first_event / users_by_first_event

# Сортируем по убыванию и выводим
print(share_returned_first_event.sort_values(ascending=False))
```

```python
# Пересчитываем количество заказов на пользователя и флаг возврата
# Считаем, сколько заказов сделал каждый user_id
orders_per_user = clean_df.groupby('user_id')['order_id'].count().reset_index() 
orders_per_user.columns = ['user_id', 'order_count']
# Присоединяем обратно к основному датасету 
clean_df = clean_df.merge(orders_per_user, on='user_id', how='left')
# Проверяем типы данных
print(f"Тип user_id в clean_df: {clean_df['user_id'].dtype}")
print(f"Тип user_id в orders_per_user: {orders_per_user['user_id'].dtype}")
# Проверяем, добавилась ли колонка и есть ли там данные
print(clean_df[['user_id', 'order_count']].head(10))
# Создаем флаг возврата: 1 если заказов >= 2, иначе 0 
clean_df['returned'] = (clean_df['order_count'] >= 2).astype(int)
```

```python
# Считаем долю возврата по типам мероприятий (для Гипотезы 1) ---
# Общее число уникальных пользователей в каждом типе мероприятия 
users_by_event = clean_df.groupby('event_type_main')['user_id'].nunique()
# Число уникальных пользователей, которые вернулись (returned == 1),
# в каждом типе # ВАЖНО: мы фильтруем датасет ПЕРЕД группировкой, чтобы считать только тех, кто вернулся
returned_users_by_event = clean_df[clean_df['returned'] == 1].groupby('event_type_main')['user_id'].nunique()
# Вычисляем долю: (вернувшиеся пользователи) / (все пользователи в сегменте) # Используем .reindex, чтобы выровнять индексы (на случай, если в каком-то сегменте не было вернувшихся) 
share_returned_event = returned_users_by_event.reindex(users_by_event.index, fill_value=0) / users_by_event

print("\n--- Доля возврата по типам мероприятий ---")
print(share_returned_event.sort_values(ascending=False))
```

```python
# Проверка Гипотезы 1 (Спорт vs Концерты) ---
sport_rate = share_returned_event.get('спорт', 0) 
concert_rate = share_returned_event.get('концерты', 0) 
overall_rate = clean_df['returned'].mean()
print(f"\n--- Проверка Гипотезы 1 ---")
print(f"Доля возврата (спорт): {sport_rate:.2%}")
print(f"Доля возврата (концерты): {concert_rate:.2%}")
print(f"Средний возврат по платформе: {overall_rate:.2%}")
if sport_rate > concert_rate: 
    print("Вывод: Гипотеза ПОДТВЕРЖДАЕТСЯ. Пользователи со спортивных мероприятий возвращаются чаще.")
else: 
    print("Вывод: Гипотеза НЕ ПОДТВЕРЖДАЕТСЯ. Пользователи с концертов возвращаются чаще или доля одинаковая.")
```

```python
# Проверка Гипотезы 2 (Топ-регионы vs Остальные) ---

# 1. Находим топ-10 регионов по числу пользователей (по количеству уникальных юзеров)
top_regions = clean_df.groupby('region_name')['user_id'].nunique().sort_values(ascending=False).head(10).index

# 2. Разделяем датасет на две группы: "Топ-10 активных" и "Все остальные"
df_top = clean_df[clean_df['region_name'].isin(top_regions)]
df_other = clean_df[~clean_df['region_name'].isin(top_regions)]

# 3. Считаем долю возврата для группы "Топ-10"
# Важно: считаем уникальных пользователей, а не строки заказов
users_top_count = df_top['user_id'].nunique()
returned_top_count = df_top[df_top['returned'] == 1]['user_id'].nunique()
share_top_regions = returned_top_count / users_top_count if users_top_count > 0 else 0

# 4. Считаем долю возврата для группы "Остальные регионы"
users_other_count = df_other['user_id'].nunique()
returned_other_count = df_other[df_other['returned'] == 1]['user_id'].nunique()
share_other_regions = returned_other_count / users_other_count if users_other_count > 0 else 0

# 5. Выводим результаты
print(f"\n--- Проверка Гипотезы 2 ---")
print(f"Количество пользователей в топ-10 регионах: {users_top_count}")
print(f"Доля возврата в топ-10 регионах: {share_top_regions:.2%}")
print(f"Количество пользователей в остальных регионах: {users_other_count}")
print(f"Доля возврата в остальных регионах: {share_other_regions:.2%}")
print(f"Средний возврат по всей платформе: {overall_rate:.2%}")

# 6. Формулируем вывод по гипотезе
if share_top_regions > share_other_regions:
    print("Вывод: Гипотеза ПОДТВЕРЖДАЕТСЯ. В самых активных регионах доля повторных заказов выше.")
else:
    print("Вывод: Гипотеза НЕ ПОДТВЕРЖДАЕТСЯ. В менее активных регионах доля возврата такая же или даже выше.")
```

```python
# Считаем среднее количество билетов на пользователя
#Нам нужно для каждого user_id посчитать среднее значение tickets_count по всем его заказам.

# Агрегируем данные по пользователям
user_stats = clean_df.groupby('user_id').agg(
    order_count=('order_id', 'nunique'),
    avg_tickets_count=('tickets_count', 'mean'),
    avg_revenue=('revenue_rub', 'mean')
).reset_index()

# Добавляем флаг возврата (уже знакомый нам)
# 1, если заказов >= 2, иначе 0
user_stats['returned'] = (user_stats['order_count'] >= 2).astype(int)

print(user_stats.head()) 
```

```python
# Разбиваем пользователей на сегменты
#Теперь создадим колонку ticket_segment, которая будет определять, к какой группе относится пользователь
# Задаём границы сегментов
bins = [0, 2, 3, 5, float('inf')]
labels = ['1-2 билета', '2-3 билета', '3-5 билетов', '5+ билетов']

# Создаём сегмент
user_stats['ticket_segment'] = pd.cut(user_stats['avg_tickets_count'], bins=bins, labels=labels, right=True)


print(user_stats['ticket_segment'].value_counts().sort_index()) 
```

```python
# Считаем долю вернувшихся в каждом сегменте
#Теперь для каждого сегмента нужно посчитать:
#1. Общее число пользователей.
#2. Число вернувшихся (у кого returned == 1).
#3. Долю вернувшихся.

# Группируем по сегменту
segment_analysis = user_stats.groupby('ticket_segment').agg(
    total_users=('user_id', 'nunique'),
    returned_users=('returned', 'sum')
).reset_index()

# Считаем долю
segment_analysis['return_rate'] = segment_analysis['returned_users'] / segment_analysis['total_users']

# Сортируем по названию сегмента для красоты
segment_analysis = segment_analysis.sort_values('ticket_segment')

print(segment_analysis) 
```

- Рассчитайте среднее время между заказами для двух групп пользователей:
    - совершившие 2–4 заказа;
    - совершившие 5 и более заказов.

```python
# Считаем интервалы между заказами для каждого пользователя

df_for_intervals = clean_df[['user_id', 'order_dt']].copy()

# Сортируем по пользователю и дате (это критически важно!)
df_for_intervals = df_for_intervals.sort_values(['user_id', 'order_dt'])

# Считаем разницу между текущей и предыдущей датой для каждого пользователя
df_for_intervals['days_diff'] = df_for_intervals.groupby('user_id')['order_dt'].diff().dt.days
```

```python
# Агрегируем данные до уровня пользователя
# Теперь для каждого пользователя считаем: 
#Сколько всего заказов (order_count). 
# Средний интервал между его заказами (avg_interval_days).

user_intervals = df_for_intervals.groupby('user_id').agg(
    order_count=('order_dt', 'count'),
    avg_interval_days=('days_diff', 'mean')  
).reset_index()

# Добавим флаг возврата (для контекста)
user_intervals['returned'] = (user_intervals['order_count'] >= 2).astype(int)

print(user_intervals.head()) 
```

```python
# Разделяем на две группы (как в условии)
#Группа А: 2–4 заказа.
#Группа Б: 5+ заказов.

group_2_4 = user_intervals[(user_intervals['order_count'] >= 2) & (user_intervals['order_count'] <= 4)]
group_5_plus = user_intervals[user_intervals['order_count'] >= 5]

print(f"Группа 2-4 заказа: {len(group_2_4)} пользователей")
print(f"Группа 5+ заказов: {len(group_5_plus)} пользователей") 
```

```python
# Считаем средний интервал для каждой группы ---

mean_interval_2_4 = group_2_4['avg_interval_days'].mean()
mean_interval_5_plus = group_5_plus['avg_interval_days'].mean()

median_interval_2_4 = group_2_4['avg_interval_days'].median()
median_interval_5_plus = group_5_plus['avg_interval_days'].median()

std_interval_2_4 = group_2_4['avg_interval_days'].std()
std_interval_5_plus = group_5_plus['avg_interval_days'].std()

print("\n--- Результаты по среднему интервалу между заказами ---")
print(f"Группа 2–4 заказа: среднее = {mean_interval_2_4:.1f} дней, медиана = {median_interval_2_4:.1f} дней, стд. отклонение = {std_interval_2_4:.1f}")
print(f"Группа 5+ заказов: среднее = {mean_interval_5_plus:.1f} дней, медиана = {median_interval_5_plus:.1f} дней, стд. отклонение = {std_interval_5_plus:.1f}")
```

Пользователи с 5+ заказами совершают покупки в среднем каждые 9.6 дней,  чем пользователи с 2–4 заказами (каждые 21.3 дня)». Короткий интервал между заказами — это надёжный индикатор высокой вовлечённости пользователя. Для задач удержания и прогнозирования оттока данный признак является одним из самых сильных: рост интервала между покупками (например, свыше 30 дней) может служить сигналом о риске потери клиента
